## Data Manipulation

### Renaming Columns

Create sample data:

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Bob"],
    "Age": [25, 30, 35, 40, 45],
    "Salary": [50000.32, 60000.56, 70000.23, 80000.12, 90000.00],
}

df = pd.DataFrame(data)
df
```

Rename all columns at once:

```python
df.columns = ["Name-new", "Age-new", "Salary-new"]
df
```

Rename an individual column:

```python
df.rename(columns={"Age-new": "Age Changed Again"}, inplace=True)
df
```

Selecting Columns by Data Type:

```python
# Select only numeric columns
df_numeric = df.select_dtypes(include=["number"])
df_numeric
```

```python
# Select only object (string) columns
df_object = df.select_dtypes(include=["object"])
df_object
```

```python
# Exclude float columns
df_no_float = df.select_dtypes(exclude=["float"])
df_no_float
```

Converting to NumPy

```python
# returns a 2D NumPy array
df_array = df.values
df_array
```

Replacing Data

```python
# Replace a single value
df.replace("Bob", "Bob New", inplace=True)
df
```

```python
# Replace multiple values
df.replace([25,40,90000.00], [10,88,1000], inplace=True)
df
```

```python
# Replace in a single column
df["Name-new"] = df["Name-new"].replace("David", "Sally")
df
```

Changing Data Types:

```python
df.dtypes
```

```python
# Convert age column to float
df["Age Changed Again"] = df["Age Changed Again"].astype(float)
df.dtypes
```
