## Pivot Tables, Melt, Stack, and Unstack

Example pivot table:

```python
data = {
    "Department": ["Sales", "Sales", "IT", "IT", "HR"],
    "Employee": ["Alice", "Bob", "Charlie", "Dave", "Eve"],
    "Salary": [50000, 60000, 70000, 80000, 45000],
}
df = pd.DataFrame(data)

pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="mean"
)
pivot
```

Melt:

```python
df = pd.DataFrame({
    "Employee": ["Alice", "Bob"],
    "Sales_Q1": [200, 150],
    "Sales_Q2": [210, 160]
})

melted = pd.melt(
    df,
    id_vars=["Employee"],
    value_vars=["Sales_Q1", "Sales_Q2"],
    var_name="Quarter",
    value_name="Sales"
)
melted
```

Stack:

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob"],
    "Math": [90, 80],
    "Science": [85, 95]
})
df
```

```python
stacked = df.set_index("Name").stack()
stacked
```

Unstack:

```python
unstacked = stacked.unstack()
unstacked
```
