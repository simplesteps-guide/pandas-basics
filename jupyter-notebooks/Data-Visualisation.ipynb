Import Pandas, NumPy & matplotlib.pyplot for the notebook:

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

---
## Visualisations

Example date:

```python
df = pd.DataFrame({
    "Product": ["A", "B", "C", "D", "E"],
    "Sales": [200, 120, 340, 300, 150],
    "Profit": [50, 20, 80, 70, 25],
    "Age": [23, 45, 36, 50, 29],
    "Height": [160, 170, 175, 180, 165],
    "Weight": [55, 70, 80, 90, 60]
})
df
```

Line Plot (default):

```python
df.plot(y="Sales", kind="line", title="Line Plot - Sales")
plt.show()
```

Bar Plot:

```python
df.plot(x="Product", y="Sales", kind="bar", title="Bar Plot - Sales")
plt.show()
```

Horizontal Bar Plot:

```python
df.plot(x="Product", y="Sales", kind="barh", title="Horizontal Bar Plot - Sales")
plt.show()
```

Histogram:

```python
df["Age"].plot(kind="hist", bins=5, edgecolor="black", title="Histogram - Age")
plt.show()
```

Box Plot:

```python
df[["Sales", "Profit"]].plot(kind="box", title="Box Plot - Sales vs Profit")
plt.show()
```

KDE / Density Plot:

```python
df["Age"].plot(kind="kde", title="KDE Plot - Age")
plt.show()
```

Area Plot:

```python
df.plot(y=["Sales", "Profit"], kind="area", alpha=0.5, title="Area Plot - Sales & Profit")
plt.show()
```

Scatter Plot:

```python
df.plot(kind="scatter", x="Height", y="Weight", title="Scatter Plot - Height vs Weight")
plt.show()
```

Pie Chart:

```python
df.set_index("Product")["Sales"].plot(kind="pie", autopct="%.1f%%", title="Pie Chart - Sales")
 # removes y-label
plt.ylabel("")
plt.show()
```
