Import Pandas for the notebook:

```python
import pandas as pd
```

Create sample data for notebook:

```python
df = pd.DataFrame({
    "Col1": ["Alice", "Bob", "Charlie", "David", "Claire"],
    "Col2": [25, 30, 35, 40, 28],
    "Col3": ["New York", "Los Angeles", "Washington", "Chicago", "Houston"],
    "Col4": [50000, 60000, 70000, 80000, 55000],
    "Col5": [5.5, 6.0, 7.0, 8.0, 5.8],
    "Col6": [True, False, True, False, True],
    "other": ["A", "B", "C", "D", "E"]
})
df
```

---
## Selecting

Single column:

```python
df["Col1"]
```

Multiple columns:

```python
df[["Col1", "Col4"]]
```

By index using .iloc[]:

```python
# first row as Series
df.iloc[0]
```

```python
# first three rows as DataFrame
df.iloc[0:3]
```

```python
# first and third rows
df.iloc[[0,2]]
```

By label using .loc[]:

```python
# row with index label 0
df.loc[0]
```

```python
# rows with index labels 0,1,2 (inclusive)
df.loc[0:2]
```

```python
# all rows, only selected columns
df.loc[:, ["Col1","Col4"]]
```

```python
# subset of rows and columns
df.loc[0:2, ["Col1","Col4"]]
```

Conditional Selection (Boolean Indexing):

```python
# rows where Col2 > 30
df[df["Col2"] > 30]
```

```python
# conditional + column selection
df.loc[df["Col2"] > 30, ["Col1","Col4"]]
```

Single cell .at[] (label-based, single element):

```python
# value at row index 0, column "Col1"
df.at[0, "Col1"]
```

Single cell .iat[] (integer position, single element):

```python
# value at row 0, column 2 (0-based position)
df.iat[0, 2]
```

Slicing - Rows:

```python
# first 3 rows (like iloc)
df[0:3]
```

Slicing - Columns:

```python
# columns Col1 to Col4 (inclusive)
df.loc[:, "Col1":"Col4"]
```

Selecting with .filter():

```python
# select specific columns
df.filter(items=["Col1","Col4"])
```

```python
# select columns containing 'Col'
df.filter(like="Col")
```
