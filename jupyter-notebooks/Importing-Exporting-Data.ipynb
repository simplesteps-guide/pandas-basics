## **Importing and Exporting Data**


### **Importing CSV Files**


Local CSV read:

```python
df = pd.read_csv('../datafiles/data.csv')
df
```

Remote CSV file:

```python
df = pd.read_csv('https://raw.githubusercontent.com/simplesteps-guide/pandas-basics/refs/heads/main/datafiles/data.csv')
df
```

CSV without headers:

```python
df = pd.read_csv('../datafiles/data_no_header.csv', header=None, names=['Name', 'Age', 'City'])
df
```

Read only selected columns:

```python
df = pd.read_csv('../datafiles/data-extra.csv', usecols=['Name','Age'])
df
```

Use a column as index:

```python
df = pd.read_csv('../datafiles/data-extra.csv', index_col='ID')
df
```

Parse dates:

```python
df = pd.read_csv('../datafiles/sales.csv', parse_dates=['OrderDate'])
df
```

Handle missing values:

```python
df = pd.read_csv('../datafiles/data2.csv', na_values=['', 'NA'])
df
```

---
### **Exporting CSV Files**

Exclude index:

```python
df.to_csv("../output_files/output_no_index.csv", index=False)
```

Change delimiter:

```python
df.to_csv("../output_files/output_semicolon.csv", sep=';')
```

Select only certain columns:

```python
df.to_csv("../output_files/output_subset.csv", columns=['Name', 'Age'], index=False)
```

Export to string instead of file:

```python
csv_string = df.to_csv(index=False)
csv_string
```

### **Importing JSON Files**

Records (list of dicts):

```python
df = pd.read_json("../datafiles/data.json", orient="records")
df
```

Index-based:

```python
df = pd.read_json("../datafiles/data2.json", orient="index")
df
```

Columns-based:

```python
df = pd.read_json("../datafiles/data3.json", orient="columns")
df
```

Line-delimited JSON (NDJSON):

```python
df = pd.read_json("../datafiles/data4.json", lines=True)
df
```

---
### Exporting JSON

Export to JSON string:

```python
json_str = df.to_json(orient="records", indent=2)
json_str
```

Export to JSON file:

```python
df.to_json("../output_files/output.json", orient="records", indent=2)
```
