## Time Series Data

Creating DatetimeIndex:

```python
index = pd.date_range(start='27/10/2019 00:00:00', periods=24, freq='h')
index
```

Creating DataFrame example:

```python
np.random.seed(3)
df = pd.DataFrame(data={'Col1':np.random.randint(low=20,high=35,size=24) },
                    index=pd.date_range(start='27/10/2019 00:00:00', periods=24,freq='h'))

df.head()
```

Accessing date parts:

```python
df['Hour'] = df.index.hour
df['Minute'] = df.index.minute
df['Seconds'] = df.index.second
df['Day'] = df.index.day
df['WeekDay_number'] = df.index.weekday
df['WeekDay_text'] = df.index.day_name()
df['Month_number'] = df.index.month
df['Month_short'] = df.index.month_name()
df['Year'] = df.index.year
df['Year-Month'] = df.index.to_period('M')

df.head()
```

Accessing date parts from non-indexed columns:

```python
df_date = pd.DataFrame(data={'Date':pd.date_range(start='27/10/2019 00:00:00', periods=24,freq='h'),
                             'Col1':np.random.randint(low=20,high=35,size=24)})
df_date.head()
```

Accessing these date parts by adding .dt after the column:

```python
df_date['Hour'] = df_date['Date'].dt.hour
df_date['Minute'] = df_date['Date'].dt.minute
df_date['Year'] = df_date['Date'].dt.year

df_date.head()
```

Converting to datetime:

```python
df = pd.DataFrame({
    "Date": ["2025-01-01", "2025-01-02", "2025-01-03"],
    "Sales": [100, 150, 200]
})

df["Date"] = pd.to_datetime(df["Date"])
df
```

Setting datetime as index:

```python
df.set_index("Date", inplace=True)
df
```

Accessing Time Series Data - by sepcific date:

```python
df.loc["2025-01-02"]
```

By date range:

```python
df.loc["2025-01-01":"2025-01-02"]
```

By year, month, or day (via string slicing):

```python
# all data in January 2025
df.loc["2025-01"]
```

```python
# all data in 2025
df.loc["2025"] 
```

Create some test data for resampling:

```python
df = pd.DataFrame(data={'Col1':np.random.randint(low=20,high=35,size=1097) },
                    index=pd.date_range(start='27/10/2019 00:00:00', end='27/10/2022 23:00:00', freq='D'))

df.head(20)
```

Frequency and Resampling

```python
# Resample daily to monthly sum
df.resample("ME").sum().head(10)
```

```python
# Resample monthly to daily mean
df.resample("D").mean().head(10)
```

Create test data for Shift and Lag demo

```python
np.random.seed(101)
df = pd.DataFrame(data={'Revenue':np.random.randint(low=200,high=350,size=15) }, 
                    index=pd.date_range(start='10/10/2020', periods=15,freq='D'))
df
```

Add columns for plus and minus 1 period shift

```python
df_example = df.copy()
df_example['RevenueShift+1'] = df_example['Revenue'].shift(periods=1) 
df_example['Revenueshift-1'] = df_example['Revenue'].shift(periods=-1) 
df_example
```

Difference:

```python
df_example = df.copy()
df_example['RevenueDiff1'] = df_example['Revenue'].diff(periods=1) 
df_example
```

Percentage Change:

```python
df_example = df.copy()
df_example['PctChange1'] = df_example['Revenue'].pct_change(periods=1) 
df_example
```

Rolling average:

```python
# 3-day rolling mean
df["Revenue_RollingMean"] = df["Revenue"].rolling(window=3).mean()
df
```
