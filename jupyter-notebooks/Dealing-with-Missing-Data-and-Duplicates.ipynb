Import Pandas and NumPy for the notebook:

```python
import pandas as pd
import numpy as np
```

Create sample data for the notebook:

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", None],
    "Age": [25, np.nan, 35, 40],
    "City": ["New York", "Los Angeles", None, "Chicago"]
})
df
```

---
## Dealing with Missing Data & Duplicates

**.isnull()**:

```python
df.isnull()
```

**.isnull().sum()**:

```python
df.isnull().sum()
```

**.dropna()**:

```python
df.dropna()
```

Reset data again:

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", None],
    "Age": [25, np.nan, 35, 40],
    "City": ["New York", "Los Angeles", None, "Chicago"]
})
df
```

**.fillna()**:

```python
df.fillna("Unknown")
```

Reset data again:

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", None],
    "Age": [25, np.nan, 35, 40],
    "City": ["New York", "Los Angeles", None, "Chicago"]
})
df
```

Set the missing age as the mean of all ages:

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

Change alice to alicia and none to unkown on the name column:

```python
df["Name"] = df["Name"].replace({ None: "Unknown", "Alice": "Alicia"})
df
```
